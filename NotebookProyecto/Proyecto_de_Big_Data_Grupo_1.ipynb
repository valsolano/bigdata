##  Antecedentes

En el mundo del consumismo en el que vivimos, la industria aérea representa uno de los sectores más dinámicos en términos de fijación de precios ya que, a diferencia de otros productos que podemos encontrar en el mercado con precios más estables, los tiquetes de avión suelen experimentar cambios significativos en cuestión de segundos. Estas inconsistencias pueden depender de múltiples factores como la demanda, competencia, costos operativos, y algunas más específicas para cada vuelo como la duración, si cuenta con escalas o la clase del servicio [2]. En el mercado aéreo de la India, particularmente en las rutas que conectan las seis principales ciudades metropolitanas (Delhi, Mumbai, Bangalore, Kolkata, Hyderabad y Chennai), se ha experimentado un crecimiento en la demanda de viajes aéreos. 
Las agencias de viajes en línea operan en un entorno altamente competitivo donde deben ofrecer a sus clientes las mejores recomendaciones de compra para mantener su lealtad y confianza. Es por esto que el desarrollo de una herramienta que pueda analizar patrones de opciones de vuelo y predecir con precisión los precios futuros se vuelve una necesidad estratégica, ya que permite a las aerolíneas optimizar la gestión de ingresos mediante tarifas dinámicas ajustadas a la demanda, la temporada y la anticipación de compra, facilitar la planificación de rutas y capacidad de flota, segmentar clientes y personalizar ofertas, todo ello convirtiendo los datos históricos en una ventaja competitiva sostenible para mejorar la experiencia del cliente y mantenerse competitivos en el mercado [3].


## Definición del problema

La compañía presentada, dedicada a la venta de boletos aéreos a través de su portal en línea, enfrenta dificultades para estimar precios de manera precisa en diferentes épocas del año. Actualmente, el método que utiliza para calcular tarifas no logra capturar adecuadamente la variabilidad de los precios, la cual depende de factores como la aerolínea, el origen y destino, la duración del vuelo, el número de escalas, la temporada y la anticipación en la compra.
El problema radica en que el portal debe procesar una gran cantidad de datos y, al no contar con un modelo predictivo eficiente, las estimaciones resultan poco confiables. Esto ha generado inconsistencias en la fijación de precios y pérdidas económicas, ya que en ocasiones los boletos se ofrecen a valores que no reflejan la realidad del mercado.


## Justificación

Implementar y entrenar un modelo de aprendizaje automático para predecir el comportamiento de los precios de los boletos aéreos permitirá aprovechar datos históricos de tarifas y variables relevantes del mercado. A partir del entrenamiento con información como precio, las aerolíneas, la ruta y el tiempo de anticipación de compra, el modelo podrá aprender patrones presentes en los datos y generar estimaciones de precios futuros. 
Este enfoque permitirá automatizar el proceso de predicción y apoyar la toma de decisiones estratégicas en la gestión de tarifas. Además, las predicciones generadas por el modelo contribuirán a optimizar las estrategias comerciales de la empresa, identificar oportunidades en rutas específicas y mejorar la competitividad en el mercado, generando valor tanto para la empresa como para los pasajeros.

## Descripción del conjunto de datos

Para este trabajo se utiliza como fuente el dataset de Kaggle “Flight Price Prediction”. Este dataset trata sobre precios de tiquetes de avión y opciones de reserva de vuelos. La información fue obtenida originalmente del sitio EaseMyTrip, que es una plataforma en línea donde las personas compran vuelos, y fue recolectada mediante web scraping utilizando la herramienta Octoparse.

En el dataset, cada fila representa una opción distinta de vuelo e incluye datos como la aerolínea, el código de vuelo, la ciudad de salida y destino, la hora de salida y llegada (agrupadas en rangos), la cantidad de escalas, la duración del viaje, los días restantes entre la compra y el vuelo, la clase del tiquete (Economy o Business) y el precio, que es la variable principal que se quiere analizar y predecir.


### Características del Dataset

El dataset consta de las siguientes columnas:  

- `airline`: almacena el nombre de la aerolínea. Es una variable categórica con 6 aerolíneas distintas.
- `flight`: guarda información del código del vuelo. Es una variable categórica.
- `source_city`: indica la ciudad de salida del vuelo. Es una variable categórica con 6 ciudades únicas.
- `departure_time`: indica la hora de salida, pero está agrupada en rangos. Es una variable categórica derivada con 6 etiquetas de tiempo.
- `stops`: representa la cantidad de escalas entre origen y destino. Es una variable categórica con 3 valores distintos.
- `arrival_time`: indica la hora de llegada, también agrupada en rangos. Es una variable categórica derivada con 6 etiquetas de tiempo.
- `destination_city`: indica la ciudad de destino donde aterriza el vuelo. Es una variable categórica con 6 ciudades únicas.
- `class`: indica la clase del tiquete. Es una variable categórica con 2 valores: Business y Economy.
- `duration`: muestra la duración total del viaje en horas. Es una variable continua.
- `days_left`: representa los días restantes entre la fecha de compra y la fecha del vuelo (se calcula como fecha del viaje - fecha de reserva). Es una variable derivada.
- `price`: es la **variable objetivo**, y almacena el precio del tiquete.

### Tamaño del Dataset

- **Instancias:** 300,153 registros.
- **Atributos:** 11 características (10 predictoras y 1 variable objetivo: Price).


### Limitaciones

- Solo tiene vuelos entre 6 ciudades principales de India, entonces no aplica para otros destinos.
- Los datos son de un rango de fechas específico (11 de febrero a 31 de marzo de 2022), así que en otras temporadas los precios pueden ser diferentes.
- La hora de salida y llegada están agrupadas por rangos y no en horas exactas, lo que reduce el detalle del análisis.
- Al venir de web scraping de una página (EaseMyTrip), los precios pueden depender de lo que el sitio mostraba en ese momento.

```python
# División estratificada por clase
strat_train_set, strat_test_set = train_test_split(dfVuelos, 
                                                     test_size=0.2, 
                                                     stratify=dfVuelos["class"], 
                                                     random_state=42)

print('Tamaño del strat_train_set', strat_train_set.shape)
print('Tamaño del strat_test_set', strat_test_set.shape)
```

```python
# Comparación de division de clases
def compare_sampling(overall, stratified, random):
    compare_props = pd.DataFrame({
        'Overall %': overall * 100,
        'Stratified %': stratified * 100,
        'Random %': random * 100
    })
    compare_props['Strat. Error %'] = ((compare_props['Stratified %'] / compare_props['Overall %']) - 1) * 100
    compare_props['Rand. Error %'] = ((compare_props['Random %'] / compare_props['Overall %']) - 1) * 100
    return compare_props

compare_props_class = compare_sampling(
    dfVuelos['class'].value_counts(normalize=True).sort_index(),
    strat_train_set['class'].value_counts(normalize=True).sort_index(),
    train_set['class'].value_counts(normalize=True).sort_index()
)

compare_props_class
```

## Nuevas columnas usando clusters

```python
from sklearn.preprocessing import FunctionTransformer
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.metrics.pairwise import rbf_kernel
from sklearn.cluster import KMeans
```

```python
class ClusterSimilarity(BaseEstimator, TransformerMixin):
    def __init__(self, n_clusters=10, gamma=1.0, random_state=None):
        self.n_clusters = n_clusters
        self.gamma = gamma
        self.random_state = random_state
 
    def fit(self, X, y=None, sample_weight=None):
        self.kmeans_ = KMeans(self.n_clusters, n_init=10,
                              random_state=self.random_state)
        self.kmeans_.fit(X, sample_weight=sample_weight)
        return self  
 
    def transform(self, X):
        return rbf_kernel(X, self.kmeans_.cluster_centers_, gamma=self.gamma)
    def get_feature_names_out(self, names=None):
        return [f"Cluster {i} similarity" for i in range(self.n_clusters)]
```

```python
cluster_simil = ClusterSimilarity(n_clusters=10, gamma=1., random_state=42)
similarities = cluster_simil.fit_transform(strat_train_set[["duration", "days_left"]],
                                           sample_weight=strat_train_set["price"])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Atributos numéricos
num_attribs = ["duration", "days_left"]

# Atributos categóricos con <= 7 categorías únicas (OneHotEncoder aplica bien)
# airline: 6, source_city: 6, departure_time: 6, stops: 3,
# arrival_time: 6, destination_city: 6, class: 2
cat_attribs_ohe = ["airline", "source_city", "departure_time", "stops",
                   "arrival_time", "destination_city", "class"]

# Pipeline numérico: imputa y estandariza
num_pipeline = Pipeline([("impute", SimpleImputer(strategy="median")),
                         ("standardize", StandardScaler())])

# Pipeline categórico (< 8 categorías): imputa y One-Hot Encodes
cat_pipeline = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)

# Integra los pipelines
preprocessing_prueba = ColumnTransformer([
    ("num", num_pipeline, num_attribs),
    ("cat", cat_pipeline, cat_attribs_ohe),
])
```

**Pipeline Definitivo**

```python
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline, Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler, OneHotEncoder, OrdinalEncoder

# Función para calcular ratios entre dos columnas
def column_ratio(X):
    return X[:, [0]] / X[:, [1]]

def ratio_name(function_transformer, feature_names_in):
    return ["ratio"]

def ratio_pipeline():
    return make_pipeline(
        SimpleImputer(strategy="median"),
        FunctionTransformer(column_ratio, feature_names_out=ratio_name),
        StandardScaler()
    )

# Pipeline logarítmico: para variables numéricas con sesgo positivo
log_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    FunctionTransformer(np.log, feature_names_out="one-to-one"),
    StandardScaler()
)

# ClusterSimilarity sobre duration y days_left
cluster_simil = ClusterSimilarity(n_clusters=10, gamma=1., random_state=42)

# Pipeline numérico por defecto
default_num_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler()
)

# Pipeline categórico 1: OneHotEncoder para variables con <= 7 categorías
# airline(6), source_city(6), departure_time(6), stops(3),
# arrival_time(6), destination_city(6), class(2)
cat_attribs_ohe = ["airline", "source_city", "departure_time", "stops",
                   "arrival_time", "destination_city", "class"]

cat_pipeline = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)

# Pipeline categórico 2: OrdinalEncoder para 'flight' (1561 categorías únicas)
# OrdinalEncoder mapea cada categoría a un entero sin generar alta dimensionalidad
cat_attribs_ord = ["flight"]

cat_pipeline2 = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)
)

# Pipeline completo
preprocessing = ColumnTransformer([
    ("ratio_dur_days", ratio_pipeline(), ["duration", "days_left"]),
    ("log", log_pipeline, ["duration", "days_left"]),
    ("geo_clusters", cluster_simil, ["duration", "days_left"]),
    ("cat", cat_pipeline, cat_attribs_ohe),
    ("cat2", cat_pipeline2, cat_attribs_ord),
],
remainder=default_num_pipeline)
```
